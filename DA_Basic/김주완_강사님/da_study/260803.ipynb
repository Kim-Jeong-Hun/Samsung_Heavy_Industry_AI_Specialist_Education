```python
# 데이터베이스 서버에 더 쉽게 접속하기 위한 라이브러리
# connect(), cursor, cursor.close(), 
# connect.close()를 하지 않아도 됨.
import sqlalchemy as sqlal
import pandas as pd

engine = sqlal.create_engine('mysql+pymysql://test:1234@121.152.203.186:33060/python_db')

emp = pd.read_sql('select * from emp', engine)
joined_emp = pd.read_sql('select * from emp join dept on emp.deptno = dept.deptno', engine)
print(joined_emp)
```

```python
# engine에 데이터베이스의 경로를 설정해두었으므로
# 'joined' 테이블을 engine 경로에 저장
# 만약 'joined_its_me' 테이블이 존재하면 대체
# 'joined' : 테이블 이름으로
# engine : engine에 설정한 경로로
# if_exists='append' : 만약 테이블이 존재한다면, 뒤에 이어 붙이기
# index=False : 앞에 붙은 0,1,2,3,4,... 등의 인덱스를 저장하지 않음.
emp.to_sql('', engine, if_exists='append', index=False)
```

```python
person = pd.read_csv('../da_data/survey_person.csv')
site = pd.read_csv('../da_data/survey_site.csv')
visited = pd.read_csv('../da_data/survey_visited.csv')
survey = pd.read_csv('../da_data/survey_survey.csv')
```

```python
# .merge() 메소드를 사용하여 데이터 연결하기
# sql로 치면 조인과 같은 개념

# == select * from site join visited on site.name = visited.name
# site와 visited를 merge (왼쪽에 name, 오른쪽에 site, 방법은 inner join)
site.merge(visited, left_on='name', right_on='site', how='inner')
```

```python
sp = survey.merge(person, left_on='person', right_on='ident')
spv = sp.merge(visited, left_on = 'taken', right_on='ident')
spvs = spv.merge(site, left_on = 'site', right_on='name')
spvs.drop(['ident_x', 'ident_y', 'name'], axis=1)
```

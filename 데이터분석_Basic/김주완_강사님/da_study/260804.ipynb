```python
# 평균 구하는 사용자 함수 정의
def my_mean(values):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    return sum/n

# 'lifeExp'에 평균 구해서 year로 묶기
agg_my_mean = df.groupby('year')['lifeExp'].agg(my_mean)
print(agg_my_mean)
```

```python
# 연도별 평균 기대수명에서 평균 기대수명 빼기
df.groupby('year')['lifeExp'].mean() - df['lifeExp'].mean()
```

```python
# 기대수명을 연도별로 묶은 후
# 평균, 표준편차, 총합(파이썬 기본함수) 반환하기
df.groupby('year')['lifeExp'].agg(['mean', 'std', sum])
```

```python
# 연도별, 대륙별 평균 기대수명
df.groupby(['year', 'continent'])['lifeExp'].mean()
```

```python
# df에서 1952년도의 Africa 대륙에 해당하는 컬럼들을 전부 반환
df[(df['year'] == 1952) & (df['continent'] == 'Africa')]
```

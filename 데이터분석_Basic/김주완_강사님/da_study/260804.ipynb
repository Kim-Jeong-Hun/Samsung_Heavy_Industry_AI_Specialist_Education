```python
# 평균 구하는 사용자 함수 정의
def my_mean(values):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    return sum/n

# 'lifeExp'에 평균 구해서 year로 묶기
agg_my_mean = df.groupby('year')['lifeExp'].agg(my_mean)
print(agg_my_mean)
```

```python
# 연도별 평균 기대수명에서 평균 기대수명 빼기
df.groupby('year')['lifeExp'].mean() - df['lifeExp'].mean()
```

```python
# 기대수명을 연도별로 묶은 후
# 평균, 표준편차, 총합(파이썬 기본함수) 반환하기
df.groupby('year')['lifeExp'].agg(['mean', 'std', sum])
```

```python
# 연도별, 대륙별 평균 기대수명
df.groupby(['year', 'continent'])['lifeExp'].mean()
```

```python
tips = sns.load_dataset('tips')
tips

hist = plt.hist(tips['total_bill'], bins=10)
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
```

```python
hist = plt.hist('total_bill', data=tips, bins=10)
print('빈도수 :', hist[0])
print('구간값 :', hist[1])
```

```python
series = []
for group in tips.groupby('sex')['total_bill']:
    series.append(group[1])

box1 = plt.boxplot(series, label=['Female', 'Male'], vert=False)
plt.title('Boxplot of Tips by Gender')
plt.xlabel('Gender')
plt.ylabel('Tip')
```

```python
s_tips = tips['sex'].value_counts()
plt.bar(s_tips.index, s_tips.values, color = ['b', 'r'], width = .5)
plt.title('Barplot of Total Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Total Bill Count')
```

```python
s_tips = tips['sex'].value_counts()
plt.pie(s_tips.values, labels = s_tips.index, autopct = '%.2f%%')
plt.title('Pie chart of Total Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Total Bill Ratio')
```

```python
plt.pie(s_tips.values, labels = s_tips.index, autopct = '%.2f%%',
startangle = 90, counterclock = False,
explode = [0.03, 0.03], shadow = True)
```

```python
def recode_gender(gender):
    if gender == 'Female':
        return 'red'
    else:
        return 'blue'
tips['color'] = tips['sex'].apply(recode_gender)
tips.head()
```

```python
plt.scatter(x = tips['total_bill'], y = tips['tip'], s = tips['size'] * 10,
c = tips['color'], alpha = 0.5)
plt.title('Total Bill vs Tip Colored by Gender and Sized by Size')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
```

```python
sns.barplot(x = 'sex', y = 'total_bill', data = tips, estimator='std', 
            errorbar=('ci', 0))
plt.title('Barplot of Total Bill by Sex')
```

```python
# 1) heidisql에서 접속
#   update, insert 변경 작업 실행!
#   commit 하지 않음.

# 2) 데이터베이스가 LOCK에 걸린 상태

import pymysql

conn = pymysql.connect(
    host='svc.sel3.cloudtype.app', user='root', password='1234', port=31776,
    db='BLEAX', charset='utf8'
)

cursor = conn.cursor()

sql = '''
SELECT * FROM 고객
'''

cursor.execute(sql)

# 결과를 리스트로 뽑아오는 듯?
result = cursor.fetchall()
for row in result:
    print(row[0], row[1], row[2], row[3])
# 0 : 고객아이디
# 1 : 이름
# 2 : 나이
# 3 : 등급
# ...


# 커서 닫고, 연결 끊기
cursor.close()
conn.close()
```

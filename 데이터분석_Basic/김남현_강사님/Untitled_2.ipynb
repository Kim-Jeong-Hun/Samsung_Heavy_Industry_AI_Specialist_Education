```python
class Computer:
    def powerOn(self):
        self.power = True
        print('전원 ON')

    def powerOff(self):
        self.power = False
        print('전원 OFF')

com1 = Computer()
com1.powerOn()

com2 = Computer()
com2.powerOff()
```

```python
# 파이썬에서 클래스 내부 메소드를 정의할 경우
# 항상 첫 번째 매개변수로 self를 적어주어야 한다.
# -> 왜? 파이썬의 규칙임

class Book:
    title = ''
    price = 0

    # 클래스 내부 메소드 정의
    def setTitle(self, title):
        self.title = title
        pass

    def setPrice(self, price):
        self.price = price
        pass

book1 = Book()
book1.setTitle('Python')
book1.setPrice(10000)
print(book1.title, book1.price)

book2 = Book()
book2.setTitle('Java')
book2.setPrice(10000)
print(book2.title, book2.price)
```

```python
class Student:
    def __init__(self):
        self.num = 1
        self.grade = 1
        self.name = 'Anonymous'

    def setNum(self, num):
        self.num = num

    def setGrade(self, grade):
        self.grade = grade

    def setName(self, name):
        self.name = name

s = Student()
print(s.num, s.grade, s.name)

s.setNum(9)
s.setGrade(3)
s.setName("Kim")
print(s.num, s.grade, s.name)
```

```python
class Product:
    def __init__(self, name, code, count):
        self.name = name
        self.code = code
        self.count = count

    def show_info(self):
        return 'name=%s, code=%s, count=%s' % (self.name, self.code, self.count)

p1 = Product('clock', 'c0001', 500)
print(p1.show_info())

p2 = Product('tv', 't0001', 1500)
print(p2.show_info())
```

```python
class Letter:
    def __init__ (self):
        self.data = ''

    def write(self, data):
        self.data += data

    def show(self):
        return self.data


letter1 = Letter()
letter1.write('가나다라')
letter1.write('카타파하')
print(letter1.show())

letter2 = Letter()
letter2.write('ABCD')
letter2.write('WXYZ')
print(letter2.show())
```

```python
class Car:
    # 클래스의 __init__ 안에 클래스 프로퍼티 생성
    def __init__(self):
        self.option = []

    def add_option(self, option):
        self.option.append(option)

    def show_option(self):
        return self.option


car1 = Car()
car2 = Car()

car1.add_option('전동 트렁크')
car1.add_option('통풍 시트')
print(car1.show_option())

car2.add_option('뒷자리 에어백')
car2.add_option('하이패스')
print(car2.show_option())
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f'{self.name} makes a noise.'

class Dog(Animal):
    def speak(self):
        return f'{self.name} barks.'

    def run(self):
        return f'{self.name} run.'

dog = Dog('ShuShu')
print(dog.speak())
print(dog.run())
```

```python
import sqlite3

# sqlite3 커넥션 풀 생성
conn = sqlite3.connect('geoje.db')

# 커넥션 커서 생성
cursor = conn.cursor()

sql = '''
    SELECT id, name, desc, mapinfo, latitude, longitude
    FROM tour
'''
try:
    cursor.execute(sql)
    result = cursor.fetchmany(10)
    for row in result:
        print(row[1], row[4], row[5])
except ValueError as e:
    print("Error :", e)
# 커서 닫고 커넥션 닫기
finally:
    cursor.close()
    conn.close()
```

```python
# 연습문제
# ● 사용자 인증 기능 작성하기
# - 조건 1) NotExistIdError, WrongPasswordError 클래스 작성
# - 조건 2) 오류 메시지를 전달할 수 있도록 생성자 작성
# - 조건 3) id/pw가 blue/1234가 아닌 경우
# 'ID를 확인해주세요', 'PW를 확인해주세요' 메시지 출력

class NotExistIdError(Exception):
    def __init__(self, msg):
        self.msg = msg

class WrongPasswordError(Exception):
    def __init__(self, msg):
        self.msg = msg


def login(id, pw):
    conditions = [
        id == 'blue',
        pw == 1234
    ]

    if not conditions[0]:
        raise NotExistIdError('ID를 확인해주세요.')

    if not conditions[1]:
        raise WrongPasswordError('PW를 확인해주세요')
    

    print('로그인 되었습니다.')


login('blue', 1234)
login('white', 1234)
login('blue', 4321)

```

```python
# ■ 연습문제
# ● 쇼핑몰 정보 및 상품 현황 출력하기
#  - 요청 URL: http://ggoreb.com/python/api/request_market.jsp
# - 요구사항
#  1) 응답 데이터 구조 확인 후 "쇼핑몰 이름"과 "전체 상품개수" 출력
#  2) 모든 상품과 가격 출력
#  3) 출력 형식에 맞게 출력

import requests
import re
url = 'http://ggoreb.com/python/api/request_market.jsp'

res = requests.get(url)
res.encoding = "None"
# print(res.text)

data = res.json()

print(f'{data["market_name"]} 쇼핑몰 상품 현황 - 총 {len(data["items"])}개')

for i in range(len(data["items"])):
    print(f'- 상품명: {data["items"][i]["title"]} | 가격: {data["items"][i]["price"]}원')
```

```python
# ■ 연습문제
# ● 전국 실시간 미세먼지 및 대기질 현황 출력하기
#  - 요청 URL: http://ggoreb.com/python/api/request_weather.jsp
# - 요구사항
#  1) 전체 지역 중 대기질이 나쁨인 지역만 필터링 후 출력
#  (state 나쁨 또는 action_required True)
#  2) 출력 형식에 맞게 출력

import requests

url = 'http://ggoreb.com/python/api/request_weather.jsp'
res = requests.get(url)
# print(res.text)

data = res.json()

print(f'[{data["data_source"]} - {data["update_time"]} 기준 위험 지역 안내]')

for i in range(len(data["regional_air_quality"])):
    if (data["regional_air_quality"][i]["status"]["state"] == '나쁨'):
        print(f'🚨 경보 지역: {data["regional_air_quality"][i]["location"]} | 미세먼지 농도: {data["regional_air_quality"][i]["pm10"]} (상태: {data["regional_air_quality"][i]["status"]["state"]})')
```

```python
# ■ 연습문제
# ● 실시간 영화 예매 차트 현황 및 평점 계산/출력하기
#  - 요청 URL: http://ggoreb.com/python/api/request_movie.jsp
# - 요구사항
#  1) 각 영화의 순위(rank), 제목(title), 예매율(booking_rate), 평점(star) 출력
#  2) 모든 영화 평점의 평균 출력
#  3) 출력 형식에 맞게 출력

import requests
url = 'http://ggoreb.com/python/api/request_movie.jsp'
res = requests.get(url)
# print(res.text)

data = res.json()

total_score = 0

print(f'[{data["theater"]} 현재 순위]')
for i in range(len(data["weekly_rank"])):
    total_score += data["weekly_rank"][i]["score"]["star"]
    print(f'[{data["weekly_rank"][i]["rank"]}위] {data["weekly_rank"][i]["title"]} | 예매율: {data["weekly_rank"][i]["booking_rate"]}% (평점: {data["weekly_rank"][i]["score"]["star"]})')
print('=========================================')
print(f'🔥 현재 상영작 평균 평점: {round(total_score/len(data["weekly_rank"]), 2)}')
```

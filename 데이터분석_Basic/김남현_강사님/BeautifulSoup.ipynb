```python
import requests
from bs4 import BeautifulSoup as bs

url = "http://ggoreb.com/python/html/static_practice.html"
res = requests.get(url)
res.encoding = 'utf-8'
soup = bs(res.text, "html.parser")

# 제목과 저자를 한꺼번에 확인
book_infos = soup.select('.book-info')

for b in book_infos:
    print(b.select("a")[0].text)
    print(b.select('span')[0].text)
```

```python
# ■ 연습문제
# ● 프로그래밍 언어 인기순위 순으로 언어명과 점유율 출력하기 - URL

import requests
from bs4 import BeautifulSoup as bs

url = 'http://ggoreb.com/python/html/tiobe-index.html'

res = requests.get(url)
res.encoding = 'utf-8'
# print(res.text)

soup = bs(res.text, "html.parser")
trs = soup.select("#tiobe-tbody > tr")
# print(trs)

for tr in trs:
    print(f'[{tr.select("td.rank")[0].text}]위 {tr.select("td.lang-name")[0].text} | 점유율: {tr.select("span.rating-badge")[0].text}')
```

```python
# ■ 연습문제
# ● 당근 중고거래 매물의 제목과 가격 출력하기 - URL

import requests
from bs4 import BeautifulSoup as bs

url = 'http://ggoreb.com/python/html/dangn.html'

res = requests.get(url)
res.encoding = 'utf-8'
# print(res.text)

# bs함수에 문자열(또는 파일)과 파서 지정
soup = bs(res.text, "html.parser")

# card 클래스의 Tag 객체 리스트 리턴
lis = soup.select(".card")
# print(lis)

for li in lis:
    print(f'{li.select(".title-line")[0].text} | {li.select(".price")[0].text}')
```

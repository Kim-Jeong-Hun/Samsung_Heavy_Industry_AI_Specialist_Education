```python
# 분류 > 10, 9 : 긍정(2) 리뷰  /  6~8 : 중립(1) / 5 이하 : 부정(0)

def score_convert(score):
  if score >= 9:
    return 2
  elif score >= 6:
    return 1
  else:
    return 0

df['sentiment'] = df['score'].apply(score_convert)
```

```python
# 비어있는 문자("''") > 누락값, 제외
df['content'] = df['content'].apply(lambda text: None if text == "''" else text)
df_na = df.dropna()
```

```python
y = df_na['sentiment']
X = df_na['content']
```

```python
import re
from konlpy.tag import Okt
p = re.compile('[^ ㄱ-ㅣ가-힣]+')
okt = Okt()

def okt_tokenizer(text):
  clean_text = p.sub('', text)
  word_pos = okt.pos(clean_text, stem = True)
  tokens = []
  for word, pos in word_pos:
    if pos != 'Josa':
      tokens.append(word)
  return tokens
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vect = TfidfVectorizer(tokenizer=okt_tokenizer, ngram_range=(1,2), max_df=0.9)
X_tfidf_vect = tfidf_vect.fit_transform(X)
```

```python
def display_topics(model, feature_names, no_top_words):
  for topic_index, topic in enumerate(model.components_):
    print('Topic #',topic_index)
    topic_word_indexes= topic.argsort()[::-1]
    top_indexes=topic_word_indexes[:no_top_words]
    feature_concat= ' '.join([feature_names[i] for i in top_indexes])
    print(feature_concat)

display_topics(lda, tfidf_vect.get_feature_names_out(), 15)
```

```python
# 텍스트를 벡터화하는게 준비 끝
import re
from konlpy.tag import Okt
p = re.compile('[^ a-zA-Zㄱ-ㅣ가-힣]+')
okt = Okt()

def okt_tokenizer(text):
  clean_text = p.sub('', text).lower()
  word_pos = okt.pos(clean_text, stem = True)
  tokens = []
  for word, pos in word_pos:
    if pos != 'Josa':
      tokens.append(word)
  return tokens
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cnt_vect = CountVectorizer(tokenizer=okt_tokenizer)
lyric_vect = cnt_vect.fit_transform(df['Lyric'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
# 빅뱅 - 붉은노을, 아이유 - 너의 의미
new_text = ['']

new_text = cnt_vect.transform(new_text)

song_sim = cosine_similarity(new_text, lyric_vect)
df['simility'] = song_sim.reshape(-1, 1)
df.sort_values('simility', ascending=False)
```

```python
import numpy as np

df = pd.read_csv('./data/rawdata_년도별 인기노래 가사.tsv', sep = '\t')
lyric_arr = np.array(df['Lyric'].values)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

embeddings = model.encode(lyric_arr)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
# 빅뱅 - 붉은노을, 아이유 - 너의 의미
new_text = ['너의 그 한 마디 말도 그 웃음도 나에겐 커다란 의미 너의 그 작은 눈빛도 쓸쓸한 뒷모습도 나에겐 힘겨운 약속 너의 모든 것은 내게로 와 풀리지 않는 수수께끼가 되네']

new_text = model.encode(new_text)

song_sim = cosine_similarity(new_text, embeddings)
df['simility'] = song_sim.reshape(-1, 1)
df.sort_values('simility', ascending=False)
```

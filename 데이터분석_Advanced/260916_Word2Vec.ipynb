```python
import nltk

nltk.download("punkt_tab")
```

```python
from sklearn.datasets import fetch_20newsgroups

train_news = fetch_20newsgroups(
    subset="train", remove=("headers", "footers", "quotes"), random_state=156
)

X_train = train_news.data
y_train = train_news.target

print(X_train[0])
```

```python
from gensim.models import Word2Vec
import nltk

tokenized = [nltk.word_tokenize(doc.lower()) for doc in X_train]

model = Word2Vec(
    sentences=tokenized, vector_size=100, min_count=5, sg=1
)  # sg=1: Skip-gram
print("어휘 수:", len(model.wv))
```

- model 안에 단어가 23296개 존재함 (5개보다 적게 사용된 단어 제거)


```python
model.wv["man"]
```

```python
# 'man'이라는 단어와 가장 유사한 단어 5개 출력
print(model.wv.most_similar("man", topn=5))
```

```python
# 'mac'이라는 단어와 가장 유사한 단어 5개 출력
print(model.wv.most_similar("mac", topn=5))
```
